```python
# to get percentage we use piechart
df['Survived'].value_counts().plot(kind='pie',autopct="%.2f")
```

```python
# Numerical Data

# a. Histogram
# Numerical data range we create bin suppose passenger age is from 0 to 80 yr and then we will create 8 bins and then we plot histogram
import matplotlib.pyplot as plt
plt.hist(df['Age'],edgecolor='black',bins=5)
```
