```python
# 1. Scatterplot (Numerical - Numerical)
# we will be using tips dataset we can clearly know that their maybe some relationship between bill and tips
sns.scatterplot(x='total_bill', y = 'tip', data = tips)
# we can see that their is some linear sort of relationship between total bill and tip
```

```python
sns.scatterplot(x='total_bill', y = 'tip', data = tips, hue = tips['sex'], style = tips['smoker'])
# this is some sort of multivariate analysis we can see that the male have generally given more tips 
# we even added the smoker or not information, we can clearly see that the blue, cross : means that it is non smoker male 
# we can also add size 
```

```python
sns.scatterplot(x='total_bill', y = 'tip', data = tips, hue = tips['sex'], style = tips['smoker'], size=tips['size'])
# bigger the size of circles more the size 
```

```python
# Boxplot ( Numerical - Categorical )
sns.boxplot(data=titanic, x = 'Sex', y = 'Age', palette='Set3')
```

```python
# Boxplot ( Numerical - Categorical )
sns.boxplot(data=titanic, x = 'Sex', y = 'Age',hue = 'Survived', palette='Set1')
# in male the younger population had more survival rate , whereas in female older population had higher survival rate
```

```python
# Distplot
# Numerical - Categorical

sns.distplot(x = titanic['Age'])
```
