Pandas Series

```python
# Pandas Series is just column and rows
my_series = [5,9,12,27]
my_var = pd.Series(my_series)
my_var
# column 
```

```python
# what if we want to pull out from here and do something
my_var[2]
```

```python
# how do we create our own heading 

# need four of them since we have four element in my_series
my_index = ["A","B","C","D"]
my_var2 = pd.Series(my_series, index = my_index)
```

```python
my_var2
# now we have our own index
```

```python
my_var2["B"]
```

```python
# my_var2[2] doesn't work anymore 
# python dict
cars = {"tesla":12,"Mercedes":42,"BMW":3}
my_var3 = pd.Series(cars)
```

```python
my_var3
# now we dont need to create a index
```

How to count data in dataframe

```python
# count distinct value- descending order
my_df['Color'].value_counts()
```

```python
# ascending
my_df['Color'].value_counts(ascending=True)
```

```python
# sometime datamissing and have no value
my_df['DogName'].value_counts()
# we can see that NaN value at row no. 16 is not listed here
```

```python
my_df['DogName'].value_counts(dropna = False)
# now we can see the length increase by one and if we decrease the dataset we can see that NaN is listed
```

```python
# get relative frequency - Percentage
my_df['Color'].value_counts(normalize = True)
```

```python
# get specific Item Count
# how many white dogs
my_df['Color'].value_counts()['WHITE']
```

```python
# count unique value - Size
my_df.groupby('Color').size()
# alphabetical order
```

```python
my_df.groupby('Color').count()
# group by count
```
