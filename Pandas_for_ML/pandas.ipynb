Pandas Series

```python
# Pandas Series is just column and rows
my_series = [5,9,12,27]
my_var = pd.Series(my_series)
my_var
# column 
```

How to make conditional Selections

```python
my_df = pd.read_csv('dog_data.csv')
my_df
```

```python
# Conditional Selection
#  > < = >= <=
# returns Booleans

my_df == "BROWN"
```

```python
# case sensitive
# return data
my_df[my_df == "BROWN"]
```

```python
# run them on a column 
my_df[my_df == "BROWN"]["Color"]
```

```python
# only color column

# want to return multiple column 
my_df[my_df == "BROWN"][["Color","Breed"]]
# whole Breed column is NaN since none of the Breec column entries have name "BROWN"
```

Multiple Conditional Selections

```python
# mulitpl conditional AND
# we want mixed brown dog
my_df[(my_df["Color"]=="BROWN") & (my_df["Breed"]=="MIXED")]
```

```python
len(my_df[(my_df["Color"]=="BROWN") & (my_df["Breed"]=="MIXED")])
```

```python
my_df[(my_df["Color"]=="BROWN") | (my_df["Breed"]=="MIXED")]
```

```python
len(my_df[(my_df["Color"]=="BROWN") | (my_df["Breed"]=="MIXED")])
```

```python
# return Specific Column
my_df[(my_df["Color"]=="BROWN") & (my_df["Breed"]=="MIXED")]["DogName"]
```

How to Fix Incomplete Data

```python
# dummy data
stuff = { 'A':[1,2,3],'B':[4,np.nan,6],'C':[7,8,9],'D':[10,11,12]}
my_df = pd.DataFrame(stuff)
my_df
```

```python
# Drop row with null data
my_df.dropna()
```

```python
my_df
```

```python
# get rid of column
# default axis = 0
my_df.dropna(axis = 1)
```

```python
# not permanent
# inplace = True to remove permanently
# set threshold
# dummy data
stuff = { 'A':[1,2,3],'B':[4,np.nan,6],'C':[7,8,9],'D':[10,11,12]}
my_df = pd.DataFrame(stuff)
my_df
```

```python
my_df.dropna(thresh=1)
```

```python
# nothing changed we want column 
my_df.dropna(thresh=2,axis=1)
```

```python
# if have two thing
# Replace thins with fillna()
```

```python
my_df.fillna(value=99)
```

```python
# use math function 
my_df.fillna(value = my_df['B'].mean())
```

```python
# use min max
my_df.fillna(value=my_df['B'].max())
```
