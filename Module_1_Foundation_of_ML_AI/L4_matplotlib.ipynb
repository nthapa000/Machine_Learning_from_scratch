```python
x = [0, 1, 2, 3, 4]
y = [2, 4, 6, 8, 10]

# plotting the line graph
```

```python
plt.plot(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("Line grapph")
plt.show()
```

```python
# Pie Chart
```

```python
labels = ['Cookies', 'Jellybean', 'Milkshake', 'Cheesecake']
sizes = [38.4, 40.6, 20.7, 10.3]
colors = ['yellowgreen', 'gold', 'lightskyblue', 'lightcoral']
```

```python
plt.pie(sizes, colors=colors)
plt.legend(labels, loc='best')
plt.tight_layout()
plt.show()
```

```python
# Scatter plot
```

```python
# taking height as x-axis and weight as y-axis, scatter the points.
```

```python
male_heights = [73, 68, 74, 71, 69, 67, 68, 68, 67, 63]
male_weights = [241, 162, 212, 220, 206, 152, 183, 167, 175, 156]
```

```python
plt.scatter(male_heights, male_weights, marker='+',label="Male")
plt.legend(loc="best")
plt.show()
```
