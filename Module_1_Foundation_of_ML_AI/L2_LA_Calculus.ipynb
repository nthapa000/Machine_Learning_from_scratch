```python
v1 = np.array([1, 0])
v2 = np.array([2, 0])
```

```python
# checking for the linear dependecy by computing the rank of the matrix formed by the vectors
matrix = np.vstack([v1,v2])
# row wise stack
rank = np.linalg.matrix_rank(matrix)
# print(rank)
is_linearly_dependent = rank < min(matrix.shape)
# print("hello")
print(is_linearly_dependent)
```

Linear Regression

```python
np.random.seed(0)
X = 2*np.random.rand(100,1)
y = 3 + 4 * X + np.random.randn(100,1)

# add a column of ones to X to account for the bias_term
X_b = np.c_[np.ones((100,1)),X]

# Compute the optimal parameters using the normal equation
theta_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
print(theta_best)
```

Visualization of Vectors using Matplotlib


```python
import numpy as np
import matplotlib.pyplot as plt

# Define the vectors
a = np.array([2, 3])
b = np.array([4, 1])

# Create the plot
plt.figure(figsize=(5, 5))

# Draw x-axis and y-axis
plt.axvline(x=0, color='grey', lw=1)
plt.axhline(y=0, color='grey', lw=1)

# Add grid
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

# Set axis limits
plt.xlim(-1, 5)
plt.ylim(-1, 5)

# Plot vector a
plt.quiver(
    0, 0, a[0], a[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='r',
    label='a'
)

# Plot vector b
plt.quiver(
    0, 0, b[0], b[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='b',
    label='b'
)

# Add legend
plt.legend()

# Show plot
plt.show()
```

Gram Schimidt Process 

```python
import numpy as np

def gram_schmidt(vectors):
    basis = []

    for v in vectors:
        # Remove projections onto existing basis vectors
        w = v - sum(np.dot(v, b) * b for b in basis)

        # Check if vector is not near zero
        if np.linalg.norm(w) > 1e-10:
            basis.append(w / np.linalg.norm(w))

    return np.array(basis)

vectors = np.array([[1, 1], [1, -1]])

orthonormal_basis = gram_schmidt(vectors)

print(orthonormal_basis)
```

How to compute the eigenvalues and eigenvectors of a matrix using NumPy and visualize them using Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[2, 1],
              [1, 2]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# origins for 2 arrows
X = [0, 0]
Y = [0, 0]

# vector components
U = eigenvectors[0, :]
V = eigenvectors[1, :]

plt.quiver(X, Y, U, V,
           color=['r', 'b'],
           scale=3)

plt.xlim(-1, 1)
plt.ylim(-1, 1)

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

plt.grid(color='gray', linestyle='--', linewidth=0.5)

plt.show()
```
