```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
# important webscrappign library
```

```python
webpage = requests.get('https://quotes.toscrape.com/page/1/').text
# <Response [200]> means we are fine 
# 403 server reqected us

# url which we will scrape
# https://quotes.toscrape.com/page/1/

# some website may reject us thinking like we are bot
# to prevent it we make them think that we are human interacting with the website using browser, we pass additional parameters like headers
```

```python
# parse and extract necessary information , here we use beautiful soup library
```

```python
soup = BeautifulSoup(webpage, 'lxml') #html parsing gets easy
```

```python
# we have BeautifulSoup object called soup and now we can use Beautiful soup fucnction
print(soup.prettify())
```

```python
# containers will repeat 
soup.find_all('a')
# print all a tags
```

```python
soup.find_all('a')[0:4][0].text
# now we are able to extract text
```

```python
for i in soup.find_all('a'):
    print(i.text)
# formatting alos added use .strip() to remove all the spaces, \n
# i dont need just tags i need other infos also i need the full quotes
```

```python
for i in soup.find_all('span'):
    print(i.text.strip())
```

```python
# extracted all the author name quote wise 
authors_name = []
for author in soup.find_all('small'):
    authors_name.append(author.text.strip())
```

```python
# trying to extract quotes
quotes = []
for quote in soup.find_all('span',class_='text'):
    quotes.append(quote.text.strip())
quotes

```

```python
authors_name
```

```python
d = {'Author name':authors_name,'Quote':quotes}
```
