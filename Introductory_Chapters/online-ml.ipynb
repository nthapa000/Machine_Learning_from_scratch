Importing libraries

```python
import numpy as np
from sklearn import linear_model
import time
```

```python
n_samples, n_features = 1, 500
y = np.random.randn(n_samples)
X = np.random.randn(n_samples, n_features)
```

```python
clf = linear_model.SGDRegressor()
# in this algorithm we have fit and partial fit alogrithm
```

```python
# recording time
start_time = time.time()
clf.partial_fit(X,y)
# thoda data deke we can partially train our model and later we can give some more data and start training from that point only
elapsed_time = time.time() - start_time
print(elapsed_time)
```
