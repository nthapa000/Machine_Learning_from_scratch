Numpy:
* is the fundamental package for scientific computing in python
* provide multiple dimension array objects and multiple derived object
* fast operations on array including mathematical, logical, shape manipulation, sorting, selection, I/O, basic linear algebra, basic statistical operation


Core of NumPy package is the nd array object.

NumPy is a C, which was created a wrapper for Python, this was done because python has very simple syntax

 Pandas, Matplotlib, sckitlearn are all made over NumPy library

* Numpy array size is fixed, unlike python list it won't grow
* need to be of same data type and occupy same size in memory, homogenous array, we dont use concept of referential array 
* famous libraries using NumPy in some level


Creating Numpy Arrays

```python
import numpy as np

# creating numpy arry
a = np.array([1,2,3])
# want list as input
print(type(a))
# vector
```

```python
# this numpy array created is a one d array
# 2D and 3D
b = np.array([[1,2,3],[4,5,6]])
# provide 2D list
print(b)
# matrix
```

```python
# similarly we can create a 3D numpy array
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(c)
# this is a tensor
```

```python
# we can make numpy array of the data type we want, till now we have seen by default numpy array of integer type

#dtype
np.array([1,2,3], dtype = float)
```

```python
#dtype
np.array([1,2,3], dtype = bool)
# 0 is treated as False ,rest are considered true
```

```python
#dtype
np.array([1,2,3], dtype = complex)
```
